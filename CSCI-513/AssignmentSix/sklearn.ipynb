## Iris Dataset

```python
import pandas as pd
data = pd.read_csv('iris.csv')
data.head()
```

```python
X = data.iloc[:, 0:4]
y = data['class']
X.head()
```

## scikit-learn

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2,random_state = 34)
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Create the model
model = LogisticRegression(penalty = None)

# Train the model
model.fit(X_train, y_train)

# Make prediction
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix, precision_score, recall_score, accuracy_score, f1_score

print('Acuraccy: ', accuracy_score(y_test, y_pred))
print('Precision: ', precision_score(y_test, y_pred, average="macro"))
print('Recall: ', recall_score(y_test, y_pred, average="macro"))
print('F1 score', f1_score(y_test, y_pred, average="macro"))
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
import plotly.express as px
cm = confusion_matrix(y_test, y_pred)
fig = px.imshow(cm)
fig.show()
```

```python
fig = px.imshow(cm, text_auto=True, aspect="auto", title='Confusion Matrix',y=[0,1,2], labels={"x":"Actual", "y":"Predicted"})
fig.show()
```

## Cross Validation

```python
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score

# prepare the cross-validation procedure
cv = KFold(n_splits=10)

# Create the model
model = LogisticRegression(penalty = None)

# Train & Evaluate model
scores = cross_val_score(model, X, y, scoring='accuracy', cv=cv)

# Print Results
print(scores)
print('Accuracy: ', sum(scores)/len(scores))
```

```python
from sklearn.model_selection import cross_validate

# Train & Evaluate model
scores = cross_validate(model, X, y, scoring=['accuracy', 'precision', 'recall', 'f1'], cv=cv)

# Print Results
print(scores)
print(scores['test_accuracy'])
```
