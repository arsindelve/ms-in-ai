```python
df = pd.read_csv('./titanic.csv')
```

## How many rows do we have?

```python
# Number of rows
print(len(df))
```

## How many columns do we have?

```python
# shape is (rows, columns)
print(df.shape[1])
```

## Which columns are numeric in value?
`describe()` only summarizes numeric columns by default, so its `columns` are the numeric ones. Note: `age` and `body` also contain numbers, but pandas reads them as text because missing values are stored as `?`.

```python
# describe() keeps only numeric columns
df.describe().columns
```

## How many people were traveling to New York city?
This counts passengers whose `home.dest` is exactly `New York, NY`.

```python
# Count of the exact 'New York, NY' destination
print(df['home.dest'].value_counts()['New York, NY'])
```

## Which fare was bought the most?

```python
# Most frequent fare
print(df['fare'].value_counts().head(1))
```

## What is the highest paid fare?

```python
# Highest fare
print(df["fare"].max())
```

## How many females survived the crash?

```python
# Female survivors
print(len(df[(df['survived'] == 1) & (df['sex'] == 'female')]))
```

## How many males survived the crash?


```python
# Male survivors
print(len(df[(df['survived'] == 1) & (df['sex'] == 'male')]))
```

## How many survivors purchased a 2nd class ticket

```python
# 2nd-class survivors
print(len(df[(df['survived'] == 1) & (df['pclass'] == 2)]))
```

## Draw a pie plot to show the age count for the survivors.
Survivors have 70 distinct ages, which makes an unreadable pie, so this shows the 25 most common ages.

```python
# Top 25 survivor ages, missing ('?') excluded
# head(25) because all 70 distinct ages make an unreadable pie
df[(df['age'] != '?') & (df['survived'] == 1)]["age"].value_counts().head(25).plot(kind='pie')
```

## Draw a bar plot of the count of people in each boat. Do not include in the plot the '?' value.

```python
df[(df['boat'] != '?')]["boat"].value_counts().plot(kind='bar')
```
