## 5. Using Plotly, visualize the relationship between car prices and whether the car is sold or not.

```python
price_order = ["low", "med", "high", "vhigh"]

fig = px.histogram(df,
                   x="RequestedPrice",
                   color="Buy",
                   barmode="stack",
                   category_orders={"RequestedPrice": price_order, "MaintenancePrice": price_order},
                   labels={"RequestedPrice": "Requested Price", "Buy": "Bought"})
fig.update_layout(yaxis_title="Number of Cars")
fig.show()
```

## 6. Using Plotly, and from the cars that were only sold, use a bar diagram to visualize the relationship between the requested price and maintenance price.

```python
fig = px.histogram(df[df['Buy'] == "yes"],
                   x="RequestedPrice",
                   color="MaintenancePrice",
                   barmode="stack",
                   category_orders={"RequestedPrice": price_order, "MaintenancePrice": price_order},
                   labels={"RequestedPrice": "Requested Price", "MaintenancePrice": "Maintenance Price"})
fig.update_layout(yaxis_title="Number of Cars")
fig.show()
```

## 7. Using Plotly, and from the cars that were only sold, visualize the relationship between car safety and maintenance price.

```python
safety_order = ["med", "high"]
fig = px.histogram(df[df['Buy'] == "yes"],
                   x="MaintenancePrice",
                   color="Safety",
                   barmode="stack",
                   category_orders=({"MaintenancePrice": price_order, "Safety": safety_order}),
                   labels={"MaintenancePrice": "Maintenance Price"})
fig.update_layout(yaxis_title="Number of Cars")
fig.show()
```

## 8. Find any other interesting relationship in the data, and visualize it using Plotly.

```python
fig = px.histogram(df,
                   x="Safety",
                   color="Buy",
                   barmode="stack",
                   category_orders={"Safety": ["low", "med", "high"]},
                   labels={"Buy": "Bought"})
fig.update_layout(yaxis_title="Number of Cars")
fig.show()
```

The relationship I found was between safety and whether or not the car was purchased. Not one single car with low safety was bought. About 31% of cars with medium safety were bought, and just under half (47%) of cars with high safety were bought. Every safety level has the same mix of the other features (price, maintenance, doors, capacity, and trunk size), so the difference comes from safety itself. From this we can conclude that the company is more likely to buy a safer car, and that low safety rules a car out no matter what else it offers.

## 9. Create two variables called X and y, where y stores all the class labels, while X stores all the class features not including the class label.

```python
X = df.iloc[:, 0:6]
y = df["Buy"]
X.head()
```

10. Next, we want to convert the data in X from string to integer. We are going to use a library
called category_encoders to do that. First, let us install this library:
pip install category_encoders <br />

`from category_encoders import OrdinalEncoder

```python
from category_encoders import OrdinalEncoder
encoder = OrdinalEncoder()
X = encoder.fit_transform(X)
```

## 11. Divide the dataset into training and test sets.
The training set should be 70% while the test set should 30%. Use 53 as the random state. How many rows does the training set have, and how many rows does the testing set have?

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3,random_state = 53)

print("Train has", X_train.shape[0], "rows")
print("Test has", X_test.shape[0], "rows")
```

## 12. Create a logistic regression classifier, then train and test your classifier.
What is the accuracy, precision, recall and f1-score of the model on the test data?

```python
model = LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print('Accuracy: ', accuracy_score(y_test, y_pred))
print('Precision: ', precision_score(y_test, y_pred, pos_label="yes"))
print('Recall: ', recall_score(y_test, y_pred, pos_label="yes"))
print('F1 score:', f1_score(y_test, y_pred, pos_label="yes"))

print('--------------')
print('Confusion Matrix')
print('--------------')

cm = confusion_matrix(y_test, y_pred)
print(cm)
```

## 13. What are the true positives, true negatives, false positives, and false negatives?


```python
print('True Positives: ', cm[1,1])
print('True Negatives: ', cm[0,0])
print('False Positives: ', cm[0,1])
print('False Negatives:', cm[1,0])
```

## 14. Although we did not learn about Support Vector Machine (SVM), let us use this classifier.

Different from logistic regression that only uses linear equation, SVM supports different types
of equations, including linear equation. These equations in SVM are called kernels. RBF
(Radial Basis Function) is one of the kernels that is a popular in classification. RBF requires
one parameter called gamma that we need to specify in Sklearn. Gamma should have a value
from 0 to 1, where a higher value will cause the SVM to fit (memorize) the training dataset,
which we do not want to do. Usually, a gamma value of 0.1 is usually used. The value of
gamma needs to be manually specified in the learning algorithm. Create an SVM model with
an RBF kernel. Then, train and test your classifier. What is the accuracy, precision, recall, f1-
score and confusion matrix of the model?

```python
model = SVC(kernel="rbf", gamma=0.1)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print('Accuracy: ', accuracy_score(y_test, y_pred))
print('Precision: ', precision_score(y_test, y_pred, pos_label="yes"))
print('Recall: ', recall_score(y_test, y_pred, pos_label="yes"))
print('F1 score:', f1_score(y_test, y_pred, pos_label="yes"))

print('--------------')
print('Confusion Matrix')
print('--------------')

cm = confusion_matrix(y_test, y_pred)
print(cm)
```

## 15. Similar to SVMs, we also did not learn about random forest as a classifier but let us try it out.
Create a random forest model with 100 trees, the number of trees can be specified as a
parameter in Sklearn. Then, train and test your classifier. What is the accuracy, precision, recall,
f1-score and confusion matrix of the model?

```python
model = RandomForestClassifier(n_estimators=100, random_state=53)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print('Accuracy: ', accuracy_score(y_test, y_pred))
print('Precision: ', precision_score(y_test, y_pred, pos_label="yes"))
print('Recall: ', recall_score(y_test, y_pred, pos_label="yes"))
print('F1 score:', f1_score(y_test, y_pred, pos_label="yes"))

print('--------------')
print('Confusion Matrix')
print('--------------')

cm = confusion_matrix(y_test, y_pred)
print(cm)
```
